```python
import pandas as pd
df = pd.read_csv("data.csv")

# Calcular o preço médio de cada local
media= df.groupby("district")["total"].mean()

# Mostrar todos os locais ordenados do menor para o maior preço
print(media.sort_values())

# Local com o maior preço médio
maior = media.idxmax()
print("\nLocal com maior preço médio:", maior)
print("Preço médio:", media.max())

# Local com o menor preço médio
menor = media.idxmin()
print("\nLocal com menor preço médio:", menor)
print("Preço médio:", media.min())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Ler a planilha
df = pd.read_csv("data.csv")

# Criar pasta para salvar os gráficos
os.makedirs("graficos", exist_ok=True)

# Gráfico 1: distribuição dos aluguéis
plt.figure(figsize=(10, 5))

plt.hist(df["rent"].dropna(), bins=30)

plt.title("Distribuição dos preços de aluguel")
plt.xlabel("Valor do aluguel (R$)")
plt.ylabel("Quantidade de imóveis")

plt.tight_layout()
plt.savefig("graficos/grafico1.png")
plt.show()
```

```python
# Gráfico 2: preço médio por distrito

# Calcular média e quantidade de imóveis
distritos = df.groupby("district")["rent"].agg(
    ["mean", "count"]
)

# Considerar distritos com pelo menos 30 imóveis
distritos = distritos[distritos["count"] >= 30]

# Mostrar os 10 distritos com maiores médias
top10 = distritos.nlargest(10, "mean").sort_values("mean")

plt.figure(figsize=(10, 6))

plt.barh(top10.index.astype(str), top10["mean"])

plt.title("10 distritos com maiores aluguéis médios")
plt.xlabel("Aluguel médio (R$)")
plt.ylabel("Distrito")

plt.tight_layout()
plt.savefig("graficos/grafico2.png")
plt.show()
```

```python
# Gráfico 3: área versus aluguel

dados = df[["area", "rent"]].dropna()

plt.figure(figsize=(10, 5))

plt.scatter(
    dados["area"],
    dados["rent"],
    alpha=0.4
)

plt.title("Relação entre área e preço do aluguel")
plt.xlabel("Área do imóvel (m²)")
plt.ylabel("Valor do aluguel (R$)")

plt.tight_layout()
plt.savefig("graficos/grafico3.png")
plt.show()
```
